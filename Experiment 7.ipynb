```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset=pd.read_csv("/home/oem/Downloads/Salary_dataset.csv")
```

```python
dataset.head()
```

```python
# Explicitly pick column 0 as X (reshaped to 2D) and column 1 as y (1D)
X=dataset.iloc[:, 0].values.reshape(-1,1)  # Shape will be (N, 1)
```

```python
y=dataset.iloc[:, 1].values  # Shape will be (N,)
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=30)
```

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
```

```python
regressor .fit(X_train,y_train)
```

```python
y_pred=regressor.predict(X_test)
```

```python
plt.scatter(X_train,y_train,color='red')
plt.plot(X_train,regressor.predict(X_train),color="blue")
```
